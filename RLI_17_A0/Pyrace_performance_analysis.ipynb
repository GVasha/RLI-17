# IE Pyrace - Performance Analysis

```python
VERSION_NAME = 'QT_v02' # the name for our model
episode = 5000
episode = 35000
QTABLE =f'models_{VERSION_NAME}/q_table_{episode}.npy'
MEMORY =f'models_{VERSION_NAME}/memory_{episode}.npy'
```

```python
#import required libraries
import numpy as np
import pandas as pd

last = 0
ep = 0
episodes = []
data = np.load(MEMORY,allow_pickle=True) # loading Object arrays
for i in range(data.shape[0]):
    if data[i,4]:
        ep += 1
        # print(ep,last,i,i-last,data[i])
        episodes.append([ep,i-last,data[i][2]])
        last = i

df = pd.DataFrame(data=episodes, index=None, columns=['episode','steps','reward'])   
# df
```

```python
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
yhat0 = savgol_filter(df.reward, int(episode/120), 3) # window size 50, polynomial order 3
yhat1 = savgol_filter(df.reward, int(episode/30), 1)

plt.figure(figsize=(16,6))
# plt.plot(df.episode,df.reward)
plt.plot(df.episode,yhat0)
plt.plot(df.episode,yhat1)
plt.title('AVERAGED REWARDS')
plt.show()

plt.figure(figsize=(16,6))
df['z'] = np.where(df.reward > 0, (df.reward)/df.steps, (10000+df.reward)/df.steps)
plt.plot(df.z.rolling(window=int(episode/15)).mean())
plt.title('AVERAGED REWARD PER ACTION STEP')
plt.show()
```

### Tests of different forms of aggregation and scaling

```python
import numpy as np
import scipy
from sklearn import preprocessing

np.set_printoptions(precision=3)
data = np.load(QTABLE,allow_pickle=True) # loading Object arrays
print(data.shape)
print('sparsity',F'{np.count_nonzero(data == 0)/data.size*100:.2f}')

# axes 0..4 bucket values in each of the radar directions, axis=5 the 3 possible actions

s0 = data.sum(axis=4).sum(axis=3).sum(axis=2).sum(axis=1)/11/11/11/11
s1 = data.sum(axis=4).sum(axis=3).sum(axis=2).sum(axis=0)/11/11/11/11
s2 = data.sum(axis=4).sum(axis=3).sum(axis=1).sum(axis=0)/11/11/11/11
s3 = data.sum(axis=4).sum(axis=2).sum(axis=1).sum(axis=0)/11/11/11/11
s4 = data.sum(axis=3).sum(axis=2).sum(axis=1).sum(axis=0)/11/11/11/11
print('s0',s0.shape)

def softmaxv(vec):
    exp = np.exp(vec)
    prob = exp / np.sum(exp)
    return prob

# Examples with uni-dimensional tensor
print('EXAMPLES WITH UNI-DIMENSIONAL TENSOR')
z = s0
print('DATA',z[0])
z_norm = preprocessing.normalize(z[0].reshape(1, -1)) # equivalent to normalize([s[0]])

"""
Reshape your data either using 
array.reshape(-1, 1) if your data has a single feature or 
array.reshape(1, -1) if it contains a single sample.
"""
s_minmax = preprocessing.MinMaxScaler()
z_minmax = s_minmax.fit_transform(z[0].reshape(-1, 1)).reshape(1, -1)
s_stdsc = preprocessing.StandardScaler()
z_stdsc = s_stdsc.fit_transform(z[0].reshape(-1, 1)).reshape(1, -1)

print('NORMALIZE',z_norm)
print('MINMAX SCALER',z_minmax)
print('STD SCALER',z_stdsc)
print('softmaxv',softmaxv(z[0]))
print('scipy',scipy.special.softmax(z[0]))

print('EXAMPLES WITH TWO-DIMENSIONAL TENSOR')
print('DATA\n',z)
z_norm = preprocessing.normalize(z)
print('NORMALIZE\n',z_norm)
print('scipy\n',scipy.special.softmax(z_norm,axis=1))
z_minmax = s_minmax.fit_transform(z.T).T
print('MINMAX\n',z_minmax)
print('scipy\n',scipy.special.softmax(z_minmax,axis=1))
z_stdsc = s_stdsc.fit_transform(z.T).T
print('standard\n',z_stdsc)
print('scipy\n',scipy.special.softmax(z_stdsc,axis=1))
```

### Average aggregation + MinMax Scaling + Softmax

```python
# TRANSFORM ALL DIRECTION TENSORS
sz = np.array([scipy.special.softmax(s_minmax.fit_transform(s0.T).T,axis=1),
               scipy.special.softmax(s_minmax.fit_transform(s1.T).T,axis=1),
               scipy.special.softmax(s_minmax.fit_transform(s2.T).T,axis=1),
               scipy.special.softmax(s_minmax.fit_transform(s3.T).T,axis=1),
               scipy.special.softmax(s_minmax.fit_transform(s4.T).T,axis=1)])
```

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
import math
from random import random

α=0
β=0
γ=0
ρ=0

xs = []
ys = []

for a in range(5): # 0 is rigth... 4 is left
    α = (90+(a-2)*360/5)*math.pi/180
    xs.append(0)
    ys.append(0)
    for r in range(11):
        ρ = r + 2
        x0 = ρ*math.cos(α)
        y0 = ρ*math.sin(α)
        xs.append(x0)
        ys.append(y0)
        for b in [1,0,2]:
            g = -1 if b==2 else b
            β = -45*g*math.pi/180
            x = x0 + 3 * math.cos(α+β) * sz[a][r][b] * 2
            y = y0 + 3 * math.sin(α+β) * sz[a][r][b] * 2
            xs.append(x)
            ys.append(y)
        xs.append(x0)
        ys.append(y0)

plt.figure(figsize=(12,12))        
plt.plot(xs, ys)
plt.title('POLICY REPRESENTATION')
plt.show()
```

## Part 3; DQN v03 Performance Analysis (Pyrace-v3)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
import os, glob

VERSION_V3 = 'DQN_v03'
MODELS_DIR_V3 = f'models_{VERSION_V3}'

# Locate the latest saved memory file for v03
memory_files_v3 = sorted(
    glob.glob(os.path.join(MODELS_DIR_V3, 'memory_*.npy')),
    key=lambda f: int(''.join(filter(str.isdigit, os.path.basename(f)))) if any(c.isdigit() for c in os.path.basename(f)) else 0
)

if memory_files_v3:
    MEMORY_V3 = memory_files_v3[-1]
    last_episode_v3 = int(''.join(filter(str.isdigit, os.path.basename(MEMORY_V3))))
    print(f'Loading {MEMORY_V3}  (episode {last_episode_v3})')
    data_v3 = np.load(MEMORY_V3, allow_pickle=True)
    print(f'Memory shape: {data_v3.shape}')
else:
    print(f'No memory files found in {MODELS_DIR_V3}/. Run Pyrace_RL_DQN_v3.py first.')
    data_v3 = None
```

```python
def memory_to_episodes(data):
    rows = []
    ep_steps = 0
    ep_reward = 0.0
    for row in data:
        ep_steps += 1
        ep_reward += float(row[2])
        if row[4]:  # done flag
            rows.append({'steps': ep_steps, 'reward': ep_reward})
            ep_steps = 0
            ep_reward = 0.0
    return pd.DataFrame(rows)

if data_v3 is not None:
    df_v3 = memory_to_episodes(data_v3)
    print(f'Episodes in memory: {len(df_v3)}')
    print(df_v3.describe())
```

```python
if data_v3 is not None:
    WINDOW_SHORT = 100
    WINDOW_LONG  = 500

    rewards = df_v3['reward'].values
    steps   = df_v3['steps'].values

    # Reward per step
    rps = np.where(rewards >= 0, rewards / steps, (10000 + rewards) / steps)

    fig, axes = plt.subplots(3, 1, figsize=(14, 12))
    fig.suptitle(f'DQN v03 — Learning Curves ({last_episode_v3} episodes)', fontsize=14)

    ax = axes[0]
    ax.plot(rewards, alpha=0.25, label='raw')
    if len(rewards) > WINDOW_SHORT:
        ax.plot(savgol_filter(rewards, min(WINDOW_SHORT | 1, len(rewards) // 2 * 2 - 1), 3),
                label=f'smoothed (w={WINDOW_SHORT})')
    if len(rewards) > WINDOW_LONG:
        ax.plot(savgol_filter(rewards, min(WINDOW_LONG | 1, len(rewards) // 2 * 2 - 1), 3),
                label=f'smoothed (w={WINDOW_LONG})')
    ax.set_title('Total Reward per Episode')
    ax.set_ylabel('reward')
    ax.legend()

    ax = axes[1]
    ax.plot(rps, alpha=0.25, label='raw')
    if len(rps) > WINDOW_SHORT:
        ax.plot(savgol_filter(rps, min(WINDOW_SHORT | 1, len(rps) // 2 * 2 - 1), 3),
                label=f'smoothed (w={WINDOW_SHORT})')
    ax.set_title('Reward per Step')
    ax.set_ylabel('reward / step')
    ax.legend()

    # steps per episode 
    ax = axes[2]
    ax.plot(steps, alpha=0.25, label='raw')
    if len(steps) > WINDOW_SHORT:
        ax.plot(savgol_filter(steps, min(WINDOW_SHORT | 1, len(steps) // 2 * 2 - 1), 3),
                label=f'smoothed (w={WINDOW_SHORT})')
    ax.set_title('Steps per Episode')
    ax.set_ylabel('steps')
    ax.set_xlabel('episode')
    ax.legend()

    plt.tight_layout()
    os.makedirs(f'analysis_{VERSION_V3}', exist_ok=True)
    out = f'analysis_{VERSION_V3}/episode_{last_episode_v3}_learning_curves.png'
    plt.savefig(out, dpi=100)
    plt.show()
    print(f'Saved to {out}')
```

```python
if data_v3 is not None:
    goal_mask  = df_v3['reward'] >= 10000
    crash_mask = df_v3['reward'] < -1000
    other_mask = ~goal_mask & ~crash_mask

    print('=== DQN v03 Summary ===')
    print(f'  Total episodes   : {len(df_v3)}')
    print(f'  Goal  (lap done) : {goal_mask.sum()}  ({100*goal_mask.mean():.1f}%)')
    print(f'  Crash            : {crash_mask.sum()} ({100*crash_mask.mean():.1f}%)')
    print(f'  Timeout / other  : {other_mask.sum()} ({100*other_mask.mean():.1f}%)')
    print(f'  Mean reward      : {df_v3["reward"].mean():.1f}')
    print(f'  Best reward      : {df_v3["reward"].max():.1f}')
    print(f'  Mean steps       : {df_v3["steps"].mean():.1f}')
```
